```python
import pandas as pd
```

```python
df = pd.read_csv("skin clinic campaign.csv")

df.head()
```

```python
df.shape
```

```python
df["Response"] = df["Response_to_Campaign"].map({"Yes":1, "No":0})

df.head()
```

```python
gender_response = (
    df.groupby("Gender")["Response"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index(name="Response Rate")
)

gender_response
```

```python
df["AgeGroup"].value_counts()
```

```python
age_response = (
    df.groupby("AgeGroup")["Response"]
      .mean()
      .mul(100)
      .round(2)
      .reset_index(name="Response Rate")
)

age_response
```

```python
age_order = ["<30", "30-50", ">50"]

df["AgeGroup"] = pd.Categorical(
    df["AgeGroup"],
    categories=age_order,
    ordered=True
)
```

```python
age_response = (
    df.groupby("AgeGroup", observed=True)["Response"]
      .mean()
      .mul(100)
      .round(2)
      .reset_index(name="Response Rate (%)")
)

age_response
```

```python
df["Purchase_Last_Quarter"].value_counts()
```

```python
purchase_response = (
    df.groupby("Purchase_Last_Quarter")["Response"]
      .mean()
      .mul(100)
      .round(2)
      .reset_index(name="Response Rate")
)

purchase_response
```

```python
df["Product_Usage"] = pd.cut(
    df["Unique_Products_Purchased"],
    bins=[0, 4, 8, float('inf')],
    labels=["1-4", "5-8", ">8"]
)

df[["Unique_Products_Purchased", "Product_Usage"]].head(10)
```

```python
df["Product_Usage"].value_counts().sort_index()
```

```python
product_response = (
    df.groupby("Product_Usage", observed=True)["Response"]
      .mean()
      .mul(100)
      .round(2)
      .reset_index(name="Response Rate")
)

product_response
```
